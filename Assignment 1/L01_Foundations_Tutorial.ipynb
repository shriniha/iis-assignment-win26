   * **Statistics:** 0 failures in 100 trials only bounds the failure rate below ≈ 3 % at 95 % confidence (rule of three, $3/n$), and only for that distribution. Move the mug, change the floor, and the number no longer applies.

2. **E2 · Changed Case Prediction (Physical Bounds):**

   - **Prediction: a nearly full mug sloshes over the rim during the turn, so coffee spills on the outside of the curve.**
   
   * Every time you turn, the coffee keeps going straight for a moment and piles up against the outer wall of the mug. A half-full mug has lots of empty space above the coffee, so the pile-up never reaches the rim. A nearly full mug has almost none, so the same pile-up goes over the edge. This is why people instinctively slow down and walk stiffly when they carry a brim-full cup.

```python
#!/usr/bin/env python3
"""Embodied Cognitive Architecture
"""

from dataclasses import dataclass, field
from typing import Dict, Any, Tuple, Optional, List
import random
import time

@dataclass(frozen=True)
class GoalSpecification:
    """Formalized, concrete goal specification derived via semantic interpretation."""
    target_object: str
    destination: str
    target_volume_ml: float
    max_allowable_force_n: float
    max_slosh_wave_mm: float
    max_nav_velocity_mps: float
    time_frame_seconds: float

@dataclass
class ParameterPacket:
    """System 1 (The Proposer): Flexible sampler proposing execution parameters."""
    grasp_pose: Tuple[float, float, float]
    max_velocity_mps: float
    tilt_rate_radps: float
    source_generator: str = "stochastic_sampler"

class Proposer:
    """ To be completed """
    def __init__(self):
        pass

@dataclass
class Observation:
    """Raw, noisy sensor measurements at time t (distinct from internal belief)."""
    timestamp: float
    measured_temperature_c: float
    measured_friction: float
    eva_position: str
    room_busyness: str



@dataclass
class BeliefState:
    """The robot's current internal belief state (updated via perception)."""
    pouch_present: bool = True
    pouch_material: str = "flexible_plastic"
    fluid_viscosity_pas: float = 1.2
    vial_ready: bool = True
    camera_active: bool = True
    fallback_path_exists: bool = True
    audit_logger_enabled: bool = True
    # Real-time environmental parameters maintained internally
    ambient_temperature_c: float = 21.5
    surface_friction_coefficient: float = 0.45
    eva_position: str = "living_room"
    room_busyness: str = "medium"  # 'low', 'medium', 'high'
    preferred_cup: str = "blue_ceramic_mug"


class InterpretationSystem:
    """Uses a semantic knowledge base to map vague instructions and context 
    into precise physical goal parameters (GoalSpecification).
    """
    def __init__(self, knowledge_base: Dict[str, Any]):
        self.kb = knowledge_base

    def interpret(self, raw_utterance: str, belief: BeliefState) -> GoalSpecification:
        print(f"\n[INTERPRETATION] Analyzing instruction: '{raw_utterance}' in context of Eva (Position: {belief.eva_position}, Room Busyness: {belief.room_busyness})")
        
        cup_type = belief.preferred_cup
        cup_properties = self.kb["cups"].get(cup_type, {"max_force": 5.0, "max_wave": 2.0})
        
        busyness_factor = self.kb["room_busyness_limits"].get(belief.room_busyness, 0.5)
        max_nav_vel = 1.2 * busyness_factor
        
        is_urgent = "schnell" in raw_utterance.lower() or "fast" in raw_utterance.lower()
        time_frame = 30.0 if is_urgent else 60.0
        target_volume = 15.0 if not is_urgent else 12.0
        
        goal = GoalSpecification(
            target_object=cup_type,
            destination=belief.eva_position,
            target_volume_ml=target_volume,
            max_allowable_force_n=cup_properties["max_force"],
            max_slosh_wave_mm=cup_properties["max_wave"],
            max_nav_velocity_mps=max_nav_vel,
            time_frame_seconds=time_frame
        )
        
        print(f" -> Derived Goal: Object={goal.target_object}, MaxForce={goal.max_allowable_force_n}N, MaxNavVel={goal.max_nav_velocity_mps}m/s, TimeFrame={goal.time_frame_seconds}s")
        return goal


class SensingSystem:
    """Simulates real-time noisy hardware sensors emitting raw observations."""
    def __init__(self):
        pass

    def sense(self, true_ambient_temp: float, true_friction: float, eva_pos: str, busyness: str) -> Observation:
        noisy_temp = true_ambient_temp + random.uniform(-0.5, 0.5)
        noisy_friction = true_friction + random.uniform(-0.05, 0.05)
        
        return Observation(
            timestamp=time.time(),
            measured_temperature_c=round(noisy_temp, 2),
            measured_friction=round(noisy_friction, 2),
            eva_position=eva_pos,
            room_busyness=busyness
        )


class PerceptionSystem:
    """Estimator module that updates the internal BeliefState using raw Observations."""
    def __init__(self, belief_state: BeliefState):
        self.belief_state = belief_state

    def update(self, observation: Observation) -> None:
        self.belief_state.ambient_temperature_c = observation.measured_temperature_c
        self.belief_state.surface_friction_coefficient = observation.measured_friction
        self.belief_state.eva_position = observation.eva_position
        self.belief_state.room_busyness = observation.room_busyness


class SemanticDigitalTwin:
    """System 2 (Semantic Digital Twin): Imperfect models with sim-to-real epistemic gaps."""
    def __init__(self, belief_state: BeliefState):
        self.belief = belief_state

    def predict_properties(self, params: ParameterPacket, goal: GoalSpecification) -> Dict[str, float]:
        stiffness_coefficient = 12.5 * (1.0 + self.belief.surface_friction_coefficient * 0.1)
        predicted_force = stiffness_coefficient * (params.max_velocity_mps ** 1.5)
        
        effective_viscosity = self.belief.fluid_viscosity_pas * (1.0 - (self.belief.ambient_temperature_c - 20.0) * 0.01)
        radius = 0.05
        wave_height_m = (params.tilt_rate_radps * radius) / effective_viscosity
        predicted_wave = wave_height_m * 1000.0

        # Injecting minor epistemic simulation error (Sim-to-Real gap)
        epistemic_force_error = random.uniform(-0.2, 0.3)
        epistemic_wave_error = random.uniform(-0.1, 0.2)

        return {
            "predicted_force_n": round(max(0.0, predicted_force + epistemic_force_error), 2),
            "predicted_wave_mm": round(max(0.0, predicted_wave + epistemic_wave_error), 2),
            "predicted_nav_velocity": min(params.max_velocity_mps, goal.max_nav_velocity_mps)
        }


class WarrantGate:
    """Validates predicted properties against the interpreted GoalSpecification."""
    def __init__(self, sdt: SemanticDigitalTwin):
        self.sdt = sdt

    def validate(self, params: ParameterPacket, goal: GoalSpecification, belief: BeliefState) -> Dict[str, Any]:
        predictions = self.sdt.predict_properties(params, goal)
        
        checks = {
            "grounded": belief.vial_ready and belief.pouch_present,
            "feasible": params.max_velocity_mps > 0.0 and params.tilt_rate_radps > 0.0,
            "safe": (predictions["predicted_force_n"] <= goal.max_allowable_force_n) and 
                    (predictions["predicted_wave_mm"] <= goal.max_slosh_wave_mm) and
                    (predictions["predicted_nav_velocity"] <= goal.max_nav_velocity_mps),
            "monitorable": belief.camera_active,
            "repairable": belief.fallback_path_exists,
            "recordable": belief.audit_logger_enabled
        }
        
        all_passed = all(checks.values())
        failed_criterion = next((name for name, passed in checks.items() if not passed), None)
        
        return {
            "passed": all_passed,
            "failed_criterion": failed_criterion,
            "predictions": predictions,
            "warrant_checks": checks
        }


class ExecutionEngine:
    """Manages physical execution and constructs the epistemic audit trail."""
    def __init__(self):
        self.audit_trail: List[Dict[str, Any]] = []

    def execute_action(self, params: ParameterPacket, goal: GoalSpecification, warrant_result: Dict[str, Any]) -> Dict[str, Any]:
        print(f"\n[ACTUATOR] Executing motion in physical world: v_max={params.max_velocity_mps} m/s")
        time.sleep(0.1)
        
        # Real-world divergence check
        predictions = warrant_result["predictions"]
        actual_force = predictions["predicted_force_n"] + random.uniform(-0.3, 0.6)
        actual_wave = predictions["predicted_wave_mm"] + random.uniform(-0.2, 0.5)
        
        spill_occurred = (actual_force > goal.max_allowable_force_n) or (actual_wave > goal.max_slosh_wave_mm)
        status = "FAILURE" if spill_occurred else "SUCCESS"
        
        execution_outcome = {
            "actual_force_n": round(actual_force, 2),
            "actual_wave_mm": round(actual_wave, 2),
            "actual_volume_transferred_ml": goal.target_volume_ml if not spill_occurred else goal.target_volume_ml - 3.0,
            "spill_occurred": spill_occurred,
            "status": status
        }
        return execution_outcome

    def record_trace(self, asked_utterance: str, goal: GoalSpecification, belief: BeliefState, 
                     proposal: ParameterPacket, warrant_result: Dict[str, Any], 
                     observation: Observation, committed: bool, execution_outcome: Optional[Dict[str, Any]] = None) -> None:
        record = {
            "timestamp": time.time(),
            "asked": {
                "raw_utterance": asked_utterance,
                "interpreter_output": {
                    "target": goal.target_object,
                    "max_force_limit_n": goal.max_allowable_force_n,
                    "max_nav_velocity": goal.max_nav_velocity_mps
                }
            },
            "belief": {
                "eva_position": belief.eva_position,
                "room_busyness": belief.room_busyness,
                "ambient_temp_c": belief.ambient_temperature_c,
                "surface_friction": belief.surface_friction_coefficient
            },
            "observed": {
                "timestamp": observation.timestamp,
                "measured_temp_c": observation.measured_temperature_c,
                "measured_friction": observation.measured_friction,
                "eva_position": observation.eva_position,
                "room_busyness": observation.room_busyness
            },
            "committed": {
                "action_approved": committed,
                "proposed_parameters": {
                    "max_velocity_mps": proposal.max_velocity_mps,
                    "tilt_rate_radps": proposal.tilt_rate_radps
                },
                "warrant_verdict": warrant_result
            },
            "execution_outcome": execution_outcome
        }
        self.audit_trail.append(record)


# --- Execution Demonstration ---
if __name__ == "__main__":
    knowledge_base = {
        "cups": {
            "blue_ceramic_mug": {"max_force": 4.5, "max_wave": 1.5},
            "plastic_cup": {"max_force": 7.0, "max_wave": 3.0}
        },
        "room_busyness_limits": {
            "low": 1.0,
            "medium": 0.6,
            "high": 0.3
        }
    }

    # 1. Initialize systems
    belief_state = BeliefState(room_busyness="medium", preferred_cup="blue_ceramic_mug")
    sensing = SensingSystem()
    perception = PerceptionSystem(belief_state)
    interpreter = InterpretationSystem(knowledge_base)
    sdt = SemanticDigitalTwin(belief_state)
    warrant_gate = WarrantGate(sdt)
    executive = ExecutionEngine()
    
    print("=== STEP 1: Sensing & Perception Update ===")
    raw_observation = sensing.sense(
        true_ambient_temp=21.5, 
        true_friction=0.45, 
        eva_pos="living_room", 
        busyness="medium"
    )
    perception.update(raw_observation)
    print(f"Raw Sensor Observation : {raw_observation}")
    print(f"Updated Internal Belief : Temp={belief_state.ambient_temperature_c}°C, Friction={belief_state.surface_friction_coefficient}")
    
    print("\n=== STEP 2: Instruction Interpretation ===")
    raw_command = "Bring me my coffee fast!"
    goal = interpreter.interpret(raw_command, belief_state)
    
    print("\n=== STEP 3: System 1 Proposer Generates Parameters ===")
    proposal_v1 = ParameterPacket(
        grasp_pose=(0.45, 0.12, 0.30),
        max_velocity_mps=0.8, 
        tilt_rate_radps=0.2
    )
    print(f"Proposer suggests: v_max = {proposal_v1.max_velocity_mps} m/s")
    
    print("\n=== STEP 4: System 2 SDT & Warrant Gate Validation (Attempt 1) ===")
    warrant_v1 = warrant_gate.validate(proposal_v1, goal, belief_state)
    print(f"SDT Predictions: Nav-Vel = {warrant_v1['predictions']['predicted_nav_velocity']} m/s")
    print(f"Warrant Checks  : {warrant_v1['warrant_checks']}")
    print(f"Gate Verdict    : REJECTED (Failed on: {warrant_v1['failed_criterion']})")
    
    executive.record_trace(raw_command, goal, belief_state, proposal_v1, warrant_v1, raw_observation, committed=False)
    
    print("\n=== STEP 5: Proposer Resamples (Attempt 2) ===")
    proposal_v2 = ParameterPacket(
        grasp_pose=(0.45, 0.12, 0.30),
        max_velocity_mps=0.45, 
        tilt_rate_radps=0.08
    )

    
    warrant_v2 = warrant_gate.validate(proposal_v2, goal, belief_state)
    print(f"SDT Predictions: Force = {warrant_v2['predictions']['predicted_force_n']} N, Wave = {warrant_v2['predictions']['predicted_wave_mm']} mm")
    print(f"Warrant Checks  : {warrant_v2['warrant_checks']}")
    print(f"Gate Verdict    : APPROVED")
    
    print("\n=== STEP 6: Physical Execution & Epistemic Audit Trail ===")
    if warrant_v2["passed"]:
        final_outcome = executive.execute_action(proposal_v2, goal, warrant_v2)
        executive.record_trace(raw_command, goal, belief_state, proposal_v2, warrant_v2, raw_observation, committed=True, execution_outcome=final_outcome)
        print(f"Execution Outcome: {final_outcome}")

    print(f"\n--- COMPLETE EPISODIC AUDIT TRAIL ({len(executive.audit_trail)} Entries) ---")
    for idx, trace in enumerate(executive.audit_trail, 1):
        status = trace['committed']['action_approved']
        outcome = trace.get('execution_outcome')
        outcome_str = f" | Outcome: {outcome['status']} (Spill: {outcome['spill_occurred']})" if outcome else ""
        print(f"Entry #{idx} | Approved: {status}{outcome_str} | Command: '{trace['asked']['raw_utterance']}'")
```

* **Question 1.1:** Map the classes and methods implemented in your Python script directly onto the functional blocks of the lecture integration diagram. Explain how data flows between these components during a single cognitive decision cycle.

**ANSWER:**
| Lecture block | Code (class → method) | What flows *in* | What flows *out* |
|---|---|---|---|
| Sensors / world | `SensingSystem.sense` | true temperature, friction, Eva position, busyness (+ Gaussian-ish noise) | `Observation` (raw, noisy, timestamped) |
| Perception / state estimation | `PerceptionSystem.update` | `Observation` | in-place update of `BeliefState` |

| System 2 · Semantic Digital Twin | `SemanticDigitalTwin.predict_properties` | `ParameterPacket`, `GoalSpecification`, `BeliefState` | predicted force, wave height, nav velocity (with epistemic error) |
| Warrant Gate | `WarrantGate.validate` | predictions + goal + belief | verdict, `failed_criterion`, 6 check booleans |
| Execution | `ExecutionEngine.execute_action` | approved packet + verdict | measured outcome (force, wave, spill, status) |
| Episodic memory / audit | `ExecutionEngine.record_trace` | *all* of the above | one audit-trail record |

**One cognitive decision cycle (data flow):**

1. **Sense →** the sensors emit a noisy `Observation`; **perceive →** it is fused into the `BeliefState`. *(Observation ≠ belief: raw numbers vs. what the robot decides to believe.)*
2. **Interpret →** "Bring me my coffee fast!" + beliefs (where is Eva? how busy is the room? which cup?) become a *physical* `GoalSpecification` (e.g. max force 4.5 N, max slosh 1.5 mm, max nav speed 0.72 m/s, 30 s).
3. **Propose →** System 1 cheaply samples a `ParameterPacket`. It is *allowed to be wrong*.
4. **Predict →** System 2 (the twin) simulates the packet under the current belief and returns predicted force / wave / speed.
5. **Warrant →** the gate checks *grounded, feasible, safe, monitorable, repairable, recordable* against the goal. **Pass →** commit; **Fail →** the `failed_criterion` flows *back* to the Proposer (resample), to Eva (ask) or to a different plan (e.g. recharge).
6. **Act →** the engine executes and measures the real outcome, which differs from the prediction (sim-to-real gap).
7. **Record →** `record_trace` stores *asked / belief / observed / committed / outcome* so the episode can be audited later (Task 5).

* **Question 1.2:** Complete the proposer class based on a probabilistic model of your choice (e.g., uniform, gaussian, mixture). Justify your choice.

* **ANSWER:**

- **Model chosen:** a **mixture of experts**
- * approach speed $v \sim w\,\mathcal{N}_{[0.05,0.8]}(\mu_{bold},\sigma_{bold}) + (1-w)\,\mathcal{N}_{[0.05,0.8]}(\mu_{cautious},\sigma_{cautious})$ (*truncated* Gaussians),
- * tilt rate $\omega \sim \text{LogNormal}$ (median 0.02 rad/s), weakly coupled to speed,
- * grasp pose $\sim \mathcal{N}(\text{perceived mug pose},\ 1\text{ cm})$,
- * plus a **cross-entropy-style adaptation**: every rejection by the `safe` criterion shrinks the means, the tilt median and the weight of the "bold" expert.



* **Why this and not uniform or a single Gaussian?**

| Choice | Reason |
|---|---|
 **Mixture** for speed | The task has two *modes of behaviour* (careful vs. hurried). "Fast!" should raise the *probability of the bold mode*, not drag one global mean into a middle that is neither. A single Gaussian cannot be bimodal; uniform ignores all prior knowledge and wastes samples in the dangerous region. |
| **Truncated** Gaussians | Speeds must be positive and below the hardware limit; truncation keeps the density honest instead of clipping. |
| **Log-normal** for tilt | A rate must be $>0$ and spans orders of magnitude; a multiplicative (log-normal) error is the natural model. |
| **Gaussian** around the grasp pose | Perception noise is small and roughly symmetric: *local exploration around the best estimate*. |
| **Adaptive feedback** | System 1 is meant to be fast and *fallible*; the Warrant Gate is the safety net. If it keeps proposing what the gate keeps rejecting, the loop is wasteful. Feedback makes rejections *informative* (the experiment below counts attempts-to-accept with and without it). |

  


```python
class Proposer:
    """System 1 - a fast, *fallible* stochastic sampler.
    
        Probabilistic model (see Q1.2):
          speed  v     ~ w * TruncN(mu_bold, sd_bold) + (1-w) * TruncN(mu_cautious, sd_cautious)
          tilt   omega ~ LogNormal(median, sigma), weakly coupled to the speed
          grasp  g     ~ N(perceived mug pose, 1 cm)
        plus cross-entropy-style feedback: every 'safe' rejection shrinks the proposal.
        """
    def __init__(self, seed=None, urgency=0.0, grasp_mean=(0.45, 0.12, 0.30), grasp_sigma=0.01):
            self.rng = random.Random(seed)
            self.grasp_mean, self.grasp_sigma = grasp_mean, grasp_sigma
            self.w_bold = min(0.6, 0.15 + 0.45 * urgency)          # "fast!" raises the odds of the bold mode
            self.mu_cautious, self.sd_cautious = 0.22, 0.05
            self.mu_bold, self.sd_bold = 0.40, 0.08
            self.tilt_median, self.tilt_sigma = 0.02, 0.45
            self.v_lo, self.v_hi = 0.05, 0.80
    
    def _trunc_gauss(self, mu, sd, lo, hi):
            for _ in range(100):
                x = self.rng.gauss(mu, sd)
                if lo <= x <= hi:
                    return x
            return min(max(mu, lo), hi)
    
    def propose(self, v_override=None, tilt_override=None):
            bold = self.rng.random() < self.w_bold
            mu, sd = (self.mu_bold, self.sd_bold) if bold else (self.mu_cautious, self.sd_cautious)
            v = v_override if v_override is not None else self._trunc_gauss(mu, sd, self.v_lo, self.v_hi)
            if tilt_override is not None:
                tilt = tilt_override
            else:
                tilt = self.tilt_median * math.exp(self.rng.gauss(0.0, self.tilt_sigma)) * (1 + 0.5 * (v - 0.22))
            grasp = tuple(round(self.rng.gauss(m, self.grasp_sigma), 4) for m in self.grasp_mean)
            tag = "mixture_bold" if bold else "mixture_cautious"
            return ParameterPacket(grasp, round(v, 3), round(tilt, 4), source_generator=tag)
    
    def feedback(self, warrant_result):
            """Learn from a rejection (cross-entropy-method flavour): shrink what the gate keeps refusing."""
            if warrant_result.get("failed_criterion") == "safe":
                self.mu_bold = max(0.12, self.mu_bold * 0.85)
                self.mu_cautious = max(0.08, self.mu_cautious * 0.92)
                self.tilt_median *= 0.85
                self.w_bold *= 0.5
```


### Task 2: Non-Linear Contact Mechanics and Material Compliance
In the base script, the contact force on the flexible pouch is modeled using a simplified power-law equation. 
* **Question 2.1:** Modify the `SemanticDigitalTwin.predict_properties` method to account for a dual-stage material compliance curve. If the contact depth exceeds $0.02 \, \text{m}$, the stiffness coefficient should increase exponentially by a factor of $1.8$ due to bottoming out against a rigid internal fixture.

**ANSWER:**
- `DualStageDigitalTwin.predict_properties` replaces the single power law by a *two-stage* curve: $F = k(d)\,v^{1.5}$ with $k(d)=k_0$ for $d\le 0.02$ m and $k(d)=1.8\,k_0$ once the pouch bottoms out against the rigid internal fixture. The indentation depth is modelled as $d = 0.065\,v$ (a twin calibration constant, since the script has no depth variable). The "exponential" factor is implemented as $k(d)=k_0\,e^{\ln(1.8)\,H(d-d_c)}$ with $d_c=0.02$ m, where $H$ is a step function (0 below the threshold, 1 above), i.e. exactly $\times 1.8$.

At $v = 0.35$ m/s: $d = 22.8$ mm $> 20$ mm, so the pouch bottoms out and the force rises from 2.70 N (single-stage) to 4.87 N.



```python
import random

CRITICAL_DEPTH_M  = 0.02    # contact depth beyond which the pouch bottoms out (m)
BOTTOM_OUT_FACTOR = 1.8     # stiffness multiplier after bottoming out
DEPTH_PER_MPS     = 0.065   # assumption: contact depth (m) = 0.065 * approach speed (m/s)

class DualStageDigitalTwin(SemanticDigitalTwin):
    """Digital twin with a dual-stage compliance curve."""

    def predict_properties(self, params, goal):
        v = params.max_velocity_mps

        # Stage 1: normal stiffness
        k = 12.5 * (1.0 + self.belief.surface_friction_coefficient * 0.1)

        # Stage 2: if contact depth > 0.02 m, stiffness x 1.8 (bottoming out)
        depth = DEPTH_PER_MPS * v
        if depth > CRITICAL_DEPTH_M:
            k *= BOTTOM_OUT_FACTOR
        predicted_force = k * (v ** 1.5)

        # Fluid sloshing (unchanged)
        effective_viscosity = self.belief.fluid_viscosity_pas * (1.0 - (self.belief.ambient_temperature_c - 20.0) * 0.01)
        predicted_wave = (params.tilt_rate_radps * 0.05) / effective_viscosity * 1000.0

        # Sim-to-real error (unchanged)
        epistemic_force_error = random.uniform(-0.2, 0.3)
        epistemic_wave_error = random.uniform(-0.1, 0.2)

        return {
            "predicted_force_n": round(max(0.0, predicted_force + epistemic_force_error), 2),
            "predicted_wave_mm": round(max(0.0, predicted_wave + epistemic_wave_error), 2),
            "predicted_nav_velocity": min(v, goal.max_nav_velocity_mps),
        }

# Check: at v = 0.35 m/s the force jumps from 2.70 N (single-stage) to 4.87 N (dual-stage)
print(13.0625 * 0.35**1.5, 1.8 * 13.0625 * 0.35**1.5)   # 2.70  4.87
```


### Task 3: Advanced Fluid Sloshing and Viscosity Temperature Coupling
Fluids in laboratory automation often exhibit temperature-dependent viscosity changes.
* **Question 3.1:** Enhance the fluid dynamics model so that ambient temperature fluctuations directly alter the effective fluid viscosity using an Arrhenius-type approximation or a linear thermal degradation coefficient ($1\% \, \text{drop per} \, ^\circ\text{C}$ above $20^\circ\text{C}$).

**ANSWER:**
### Question 3.1: Viscosity that depends on ambient temperature

The slosh model predicts the wave height as $h = \dfrac{\omega\, r}{\eta_{eff}}$ (tilt rate $\omega$, container radius $r$, effective viscosity $\eta_{eff}$). In the base script the viscosity only depended on temperature through an inline formula. I moved this into its own method `effective_viscosity()` that reads the temperature from the robot's belief and offers two models:

- **Linear thermal degradation (as specified):** $\eta(T) = \eta_{20}\,[\,1 - 0.01\,(T - 20)\,]$, i.e. a 1 % drop per °C above 20 °C.
- **Arrhenius-type approximation:** $\eta(T) = \eta_{20}\,\exp\!\Big[\tfrac{E_a}{R}\Big(\tfrac{1}{T_K} - \tfrac{1}{293.15}\Big)\Big]$ with $E_a/R = 2400$ K (water-like fluid), where $T_K = T + 273.15$.

Both give $\eta_{20} = 1.2$ Pa·s at 20 °C. Because the wave height scales with $1/\eta$, a warmer room means a thinner fluid and a **bigger wave**. The Arrhenius model is more pessimistic at high temperatures (at 28 °C it predicts 0.965 Pa·s against 1.104 Pa·s for the linear model).

```python
import math, random

EA_OVER_R = 2400.0   # Arrhenius activation temperature Ea/R in kelvin (assumption: water-like fluid)

class ThermalDigitalTwin(DualStageDigitalTwin):
    """Dual-stage twin whose fluid viscosity depends on the ambient temperature."""

    def __init__(self, belief, viscosity_model="linear"):
        super().__init__(belief)
        self.viscosity_model = viscosity_model       # "linear" or "arrhenius"

    def effective_viscosity(self):
        T   = self.belief.ambient_temperature_c
        eta = self.belief.fluid_viscosity_pas        # viscosity at 20 degC
        if self.viscosity_model == "arrhenius":
            return eta * math.exp(EA_OVER_R * (1.0 / (T + 273.15) - 1.0 / 293.15))
        return eta * max(0.2, 1.0 - 0.01 * (T - 20.0))   # 1 % drop per degC above 20

    def predict_properties(self, params, goal):
        v = params.max_velocity_mps

        # contact force with dual-stage stiffness (as in Question 2.1)
        k = 12.5 * (1.0 + self.belief.surface_friction_coefficient * 0.1)
        if DEPTH_PER_MPS * v > CRITICAL_DEPTH_M:
            k *= BOTTOM_OUT_FACTOR
        predicted_force = k * v ** 1.5

        # slosh wave with temperature-dependent viscosity (new)
        predicted_wave = (params.tilt_rate_radps * 0.05) / self.effective_viscosity() * 1000.0

        return {
            "predicted_force_n": round(max(0.0, predicted_force + random.uniform(-0.2, 0.3)), 2),
            "predicted_wave_mm": round(max(0.0, predicted_wave + random.uniform(-0.1, 0.2)), 2),
            "predicted_nav_velocity": min(v, goal.max_nav_velocity_mps),
        }

# Demo: effective viscosity of the fluid [Pa.s] for both models
print(f"{'T [C]':>6} | {'linear':>7} | {'Arrhenius':>9}")
for T in (15, 20, 25, 28, 32):
    b = BeliefState(ambient_temperature_c=T)
    lin = ThermalDigitalTwin(b, "linear").effective_viscosity()
    arr = ThermalDigitalTwin(b, "arrhenius").effective_viscosity()
    print(f"{T:>6} | {lin:7.3f} | {arr:9.3f}")
```


* **Question 3.2:** Simulate an environment where the ambient temperature rises to $28^\circ\text{C}$. How does this change the predicted wave height for a tilt rate of $0.15 \, \text{rad/s}$? Does the Warrant Gate trigger a safety violation?

**ANSWER:**
### Ambient temperature rises to 28 °C (tilt rate 0.15 rad/s)

**Method:** I compare the baseline room (21.5 °C, the default belief) with a hot room (28 °C) using the temperature-dependent twin from Question 3.1 and the same packet (speed 0.25 m/s, tilt rate 0.15 rad/s). The wave height is $h = \omega r/\eta_{eff}$ with $r = 0.05$ m.

| Room | Viscosity (linear) | Wave (linear) | Viscosity (Arrhenius) | Wave (Arrhenius) |
|---|---|---|---|---|
| 21.5 °C | 1.182 Pa·s | 6.35 mm | 1.151 Pa·s | 6.52 mm |
| 28 °C | 1.104 Pa·s | 6.79 mm | 0.965 Pa·s | 7.77 mm |

**How it changes the wave height:** the hotter fluid is thinner, so the predicted wave grows by about **7 %** (linear model) or **19 %** (Arrhenius model).

**Does the Warrant Gate trigger a safety violation? Yes.** The wave height (about 6.8 mm) is far above the 1.5 mm limit of the blue mug, so the `safe` criterion fails and the packet is rejected.

**Reading the result:** the heat is not what causes the violation, because even at 21.5 °C the tilt rate of 0.15 rad/s is about four times too large. The warmth makes things worse, though: the largest tilt rate that still fits within 1.5 mm drops from 0.0355 rad/s (21.5 °C) to 0.0331 rad/s (28 °C). So the robot needs an up-to-date temperature belief to choose a safe tilt rate.


```python
import random
random.seed(5)

knowledge_base = {
    "cups": {"blue_ceramic_mug": {"max_force": 4.5, "max_wave": 1.5}},
    "room_busyness_limits": {"low": 1.0, "medium": 0.6, "high": 0.3},
}
packet = ParameterPacket(grasp_pose=(0.45, 0.12, 0.30), max_velocity_mps=0.25, tilt_rate_radps=0.15)

def wave_mm(belief, model, tilt):
    """Noiseless predicted wave height in mm: h = tilt * r / viscosity."""
    return tilt * 0.05 / ThermalDigitalTwin(belief, model).effective_viscosity() * 1000

# ---- How does 28 degC change the predicted wave height? ----------------------------------
print("Predicted wave height for tilt rate 0.15 rad/s")
waves = {}
for T in (21.5, 28.0):
    b = BeliefState(ambient_temperature_c=T)
    for model in ("linear", "arrhenius"):
        waves[(T, model)] = wave_mm(b, model, 0.15)
        print(f"  {T:>4} C  {model:<9}: {waves[(T, model)]:.2f} mm")
for model in ("linear", "arrhenius"):
    change = (waves[(28.0, model)] / waves[(21.5, model)] - 1) * 100
    print(f"  -> {model}: wave grows by {change:.1f} % when the room warms from 21.5 to 28 C")

# ---- Does the Warrant Gate raise a safety violation at 28 degC? ---------------------------
belief28 = BeliefState(ambient_temperature_c=28.0, preferred_cup="blue_ceramic_mug")
goal     = InterpretationSystem(knowledge_base).interpret("Bring me my coffee", belief28)
gate     = WarrantGate(ThermalDigitalTwin(belief28, "linear"))
verdict  = gate.validate(packet, goal, belief28)

print("\nPredicted wave (incl. sim-to-real error):", verdict["predictions"]["predicted_wave_mm"],
      "mm   | limit:", goal.max_slosh_wave_mm, "mm")
print("Warrant checks:", verdict["warrant_checks"])
print("Gate verdict  :", "APPROVED" if verdict["passed"] else f"REJECTED (failed on: {verdict['failed_criterion']})")
print("Safety violation triggered:", not verdict["warrant_checks"]["safe"])

# ---- Which tilt rate would still be safe? ------------------------------------------------
for T in (21.5, 28.0):
    eta = ThermalDigitalTwin(BeliefState(ambient_temperature_c=T), "linear").effective_viscosity()
    print(f"Largest tilt rate within 1.5 mm at {T} C: {1.5e-3 * eta / 0.05:.4f} rad/s")
```


### Task 4: Expanding the Warrant Gate Checklist
The current Warrant Gate evaluates six core criteria (*grounded, feasible, safe, monitorable, repairable, recordable*). 
* **Question 4.1:** Add a seventh criterion called **`energetically_viable`** to the Warrant Gate dictionary checks. It must evaluate whether the estimated battery drain for the requested trajectory stays below $15\%$ of total capacity.

**ANSWER**

Seventh criterion `energetically_viable`

The twin estimates the battery drain of a trajectory. The robot drives a route of length $L$ at speed $v$, so the trip lasts $t = L/v$ and the energy used is

$$E = (P_{idle} + P_{drive}\, v^2) \cdot t$$

I assume $P_{idle} = 45$ W, $P_{drive} = 60$ W·s²/m² and a 200 Wh battery. The drain in % of total capacity is $100 \cdot E / 200$ Wh. The new gate `WarrantGateV2` keeps the six original checks and adds `energetically_viable`, which is true only if

1. the estimated drain is **at most 15 %** of the total capacity, and
2. the battery still holds a **10 % reserve** after the trip. (This second condition is my design choice: a robot at 9 % must not start a trip, even if the trip itself is cheap.)

*Side note:* idle power dominates, so driving slowly is not free. A slow, careful trajectory costs more energy.

* **Question 4.2:** Introduce a low-battery state into the `BeliefState` and demonstrate how a failed `energetically_viable` check redirects the system workflow to request a recharge action rather than proceeding with the coffee delivery.

**ANSWER:**

Low-battery state and recharge redirect

`BeliefStateV2` extends `BeliefState` with `battery_pct` (the low-battery state) and `route_length_m`. The function `decide()` uses the gate's verdict to choose the next action:

- all checks pass: **deliver the coffee**;
- `energetically_viable` fails: **go to the charger**. Resampling the parameters cannot create energy, so the *plan* changes, not the packet;
- another check fails: resample a new packet.

Recharge workflow: navigate to the dock, charge to at least 40 %, interpret the request again, then deliver the coffee.




```python
from dataclasses import dataclass
import random

# ---------- belief with a battery state (Q4.2) ----------
@dataclass
class BeliefStateV2(BeliefState):
    battery_pct: float = 85.0       # state of charge in % (interoception)
    route_length_m: float = 8.0     # length of the planned trajectory

# ---------- energy model in the twin ----------
P_IDLE_W, P_DRIVE_W, CAPACITY_WH = 45.0, 60.0, 200.0
MAX_DRAIN_PCT, RESERVE_PCT = 15.0, 10.0

class EnergyAwareTwin(ThermalDigitalTwin):
    def energy_drain_pct(self, v):
        seconds = self.belief.route_length_m / max(v, 0.05)
        watts   = P_IDLE_W + P_DRIVE_W * v ** 2
        return 100.0 * (watts * seconds / 3600.0) / CAPACITY_WH

    def predict_properties(self, params, goal):
        out = super().predict_properties(params, goal)
        out["predicted_battery_drain_pct"] = round(self.energy_drain_pct(params.max_velocity_mps), 3)
        return out

# ---------- Q4.1: gate with the seventh criterion ----------
class WarrantGateV2(WarrantGate):
    def validate(self, params, goal, belief):
        result = super().validate(params, goal, belief)          # the six original checks
        drain = result["predictions"]["predicted_battery_drain_pct"]
        result["warrant_checks"]["energetically_viable"] = (
            drain <= MAX_DRAIN_PCT and (belief.battery_pct - drain) >= RESERVE_PCT)
        result["passed"] = all(result["warrant_checks"].values())
        result["failed_criterion"] = next((k for k, ok in result["warrant_checks"].items() if not ok), None)
        return result

# ---------- Q4.2: workflow decision ----------
def decide(packet, goal, belief, gate):
    verdict = gate.validate(packet, goal, belief)
    if verdict["passed"]:
        return "DELIVER_COFFEE", verdict
    if not verdict["warrant_checks"]["energetically_viable"]:    # resampling cannot fix an energy problem
        return "GO_TO_CHARGER", verdict
    return "RESAMPLE_PACKET", verdict

# ---------- demonstration ----------
random.seed(11)
knowledge_base = {"cups": {"blue_ceramic_mug": {"max_force": 4.5, "max_wave": 1.5}},
                  "room_busyness_limits": {"low": 1.0, "medium": 0.6, "high": 0.3}}
goal   = InterpretationSystem(knowledge_base).interpret("Bring me my coffee", BeliefStateV2())
packet = ParameterPacket((0.45, 0.12, 0.30), max_velocity_mps=0.25, tilt_rate_radps=0.02)   # a safe packet

scenarios = [("Healthy robot, normal trip",   dict(battery_pct=85, route_length_m=8)),
             ("LOW BATTERY (9 %)",            dict(battery_pct=9,  route_length_m=8)),
             ("Very long trajectory (900 m)", dict(battery_pct=85, route_length_m=900))]

for title, kw in scenarios:
    belief = BeliefStateV2(preferred_cup="blue_ceramic_mug", **kw)
    gate   = WarrantGateV2(EnergyAwareTwin(belief))
    action, verdict = decide(packet, goal, belief, gate)
    print(f"\n{title}")
    print("  checks   :", verdict["warrant_checks"])
    print(f"  drain    : {verdict['predictions']['predicted_battery_drain_pct']:.2f} %   battery: {belief.battery_pct} %")
    print("  decision :", action)
    if action == "GO_TO_CHARGER":
        print("  workflow : navigate to dock -> charge to >= 40 % -> re-interpret request -> deliver coffee")
```


### Task 5: Post-Hoc Epistemic Audit Analysis
A core pillar of trustworthy robotics is the ability to audit *why* a decision was made or rejected.
* **Question 5.1:** Write a post-processing function (`audit_analyzer(audit_trail)`) that parses the recorded episodic log. It should compute and print:
  1. The total number of rejected proposals versus approved executions.
  2. The most frequent warrant criterion responsible for rejection.
  3. The average variance between predicted forces and observed physical outcomes.

**ANSWER:**

`audit_analyzer(audit_trail)`

`ExecutionEngine.record_trace` writes one record per attempt. The analyzer reads this log and reports:

1. **Rejected vs. approved:** it counts the records with `committed.action_approved == False` and `True`. It also counts how many approved executions still ended in a spill (approved but wrong).
2. **Most frequent rejection reason:** it counts `committed.warrant_verdict.failed_criterion` over the rejected records.
3. **Prediction vs. reality:** for every executed record, it computes `actual_force_n - predicted_force_n`. It reports the mean (bias), the variance of this difference, and the mean squared deviation (which equals variance + bias²).

To have a log to analyse, I run 60 simulated delivery episodes with different cups, temperatures, room busyness levels and battery states.


```python
import io, contextlib, random, statistics
from collections import Counter

def audit_analyzer(audit_trail):
    """Parse the episodic log and print the audit report."""
    approved = [t for t in audit_trail if t["committed"]["action_approved"]]
    rejected = [t for t in audit_trail if not t["committed"]["action_approved"]]

    # 1) rejected vs approved
    spills = sum(1 for t in approved if t["execution_outcome"]["spill_occurred"])
    print("=" * 60 + "\n EPISTEMIC AUDIT REPORT\n" + "=" * 60)
    print(f"1) proposals logged : {len(audit_trail)}")
    print(f"   approved & executed: {len(approved)}   (of these {spills} ended in a spill)")
    print(f"   rejected           : {len(rejected)}")

    # 2) most frequent rejection reason
    reasons = Counter(t["committed"]["warrant_verdict"]["failed_criterion"] for t in rejected)
    print("2) rejection reasons:")
    for name, n in reasons.most_common():
        print(f"     {name:<22}{n:>4}  " + "#" * n)
    top = reasons.most_common(1)[0] if reasons else (None, 0)
    print(f"   -> most frequent: '{top[0]}' ({top[1]}x)")

    # 3) predicted force vs observed outcome
    diffs = [t["execution_outcome"]["actual_force_n"] - t["committed"]["warrant_verdict"]["predictions"]["predicted_force_n"]
             for t in approved]
    bias = statistics.mean(diffs)
    var  = statistics.pvariance(diffs)
    msd  = statistics.mean(d * d for d in diffs)
    print("3) predicted vs observed force (observed - predicted, in N):")
    print(f"     mean bias = {bias:+.3f} | variance = {var:.4f} | mean squared deviation = {msd:.4f} | n = {len(diffs)}")
    if bias > 0:
        print("   -> the twin under-predicts the force on average (it is too optimistic).")
    return {"approved": len(approved), "rejected": len(rejected), "top_reason": top[0], "bias": bias, "variance": var}

# ---------- generate a log: 60 simulated delivery episodes ----------
random.seed(2026)
knowledge_base = {
    "cups": {"blue_ceramic_mug": {"max_force": 4.5, "max_wave": 1.5},
             "white_ceramic_mug": {"max_force": 4.2, "max_wave": 1.2},
             "plastic_cup": {"max_force": 7.0, "max_wave": 3.0}},
    "room_busyness_limits": {"low": 1.0, "medium": 0.6, "high": 0.3},
}
belief      = BeliefStateV2()
sensing     = SensingSystem()
perception  = PerceptionSystem(belief)
interpreter = InterpretationSystem(knowledge_base)
gate        = WarrantGateV2(EnergyAwareTwin(belief))
engine      = ExecutionEngine()

for ep in range(60):
    belief.preferred_cup = random.choice(["blue_ceramic_mug", "white_ceramic_mug", "plastic_cup"])
    belief.battery_pct   = random.choice([6, 9]) if random.random() < 0.10 else random.uniform(40, 100)
    obs = sensing.sense(random.uniform(19, 32), random.uniform(0.3, 0.6),
                        random.choice(["living_room", "kitchen"]), random.choice(["low", "medium", "high"]))
    perception.update(obs)
    command = random.choice(["Bring me my coffee", "Bring me my coffee fast!"])
    with contextlib.redirect_stdout(io.StringIO()):                  # hide the interpreter printout
        goal = interpreter.interpret(command, belief)
    proposer = Proposer(seed=ep, urgency=1.0 if "fast" in command else 0.0)

    for attempt in range(6):
        packet  = proposer.propose()
        verdict = gate.validate(packet, goal, belief)
        if verdict["passed"]:
            with contextlib.redirect_stdout(io.StringIO()):
                outcome = engine.execute_action(packet, goal, verdict)
            engine.record_trace(command, goal, belief, packet, verdict, obs, True, outcome)
            break
        engine.record_trace(command, goal, belief, packet, verdict, obs, False)
        if not verdict["warrant_checks"]["energetically_viable"]:
            break                                                    # needs a recharge, not a new packet
        proposer.feedback(verdict)

# ---------- run the analysis ----------
report = audit_analyzer(engine.audit_trail)
```



* **Question 5.2:** Discuss in your lab report: *Why does maintaining separate records for `asked`, `belief`, `committed`, and `observed` solve the epistemological gap that pure data-driven imitation learning cannot address?*

**ANSWER:**

Why four separate records (`asked`, `belief`, `committed`, `observed`)?

Think of a **flight recorder, not a home video**. A video tells you *that* the plane crashed. The black box tells you which instrument said what, what the crew believed, what they decided, and what actually happened.

Pure imitation learning stores only pairs of *(what the robot saw → what the demonstrator did)*. The whole chain "meaning → belief → decision → outcome" is collapsed into one opaque mapping, so when something goes wrong it cannot say **which link broke**. Separate records restore this missing structure:

| Record | Question it answers | Failure it isolates |
|---|---|---|
| **`asked`** | What did the human mean, and how was it formalized? | Misinterpretation (wrong cup, wrong limits) |
| **`belief`** | What did the robot take to be true about the world? | Wrong or stale belief (temperature, friction, Eva's position) |
| **`observed`** | What did the sensors actually say? | Sensor noise versus belief drift (compare `observed` with `belief`) |
| **`committed`** | What was predicted, which checks passed, and why was the robot allowed to act? | Wrong prediction or a missing check |
| **outcome** | What really happened? | The gap between prediction and reality |

Because the records are separate, we can audit the system:

- **Locate the cause.** If the coffee spills, we can tell whether the interpretation, the perception, the digital twin or the world was at fault by comparing neighbouring records.
- **Measure the epistemic gap.** The analyzer shows the twin under-predicts the force by about 0.15 N. We can add that to the safety margin. A policy network never made a prediction it could be held accountable for.
- **Justify refusals.** A rejected proposal with a `failed_criterion` is evidence of why the robot correctly did *not* act. Imitation data only contains what was done.
- **Spot distribution shift early.** When `belief` leaves the range seen before, the log shows it before a failure happens.

In short, imitation learning learns correlations in behaviour, while the four-record trace keeps the **reasons**. This makes the robot's competence checkable and not only observed.

|---|---|---|---|---|

The joints are the degrees of freedom, and the links set the reach. | **Propose / Predict:** grasp poses and the tilt rate are joint-space quantities, and the twin needs the arm geometry. **Warrant:** joint limits are part of the `feasible` check. |
| 2 | **Gripper** (blue dot at the wrist) | **End effector** | Grasps the object (the cutting board in the figure, the mug in our task) and applies the grip force. | **Act:** it creates the contact force $F = k\,v^{1.5}$ that the **`safe`** check limits. |

| 4 | **Wheel unit** (blue dot at the bottom) | **Locomotion** (wheels instead of legs) | Moves the base to Eva. Wheels are efficient and smooth on flat floors, so the coffee sloshes less. | **Act:** executes the navigation speed. This is the `max_velocity_mps` that the Warrant Gate checks. |
| 5 | **Head camera** (red dot on the head, RGB-D camera) | **Exteroceptive sensor** | Sees the scene (the pot on the stove in the inset image) and measures depth. | **Sense → Perceive:** produces the `Observation` that updates the `BeliefState`. The `monitorable` check needs the camera to be active. |
| 6 | **Laser scanner** (red dot on the base, 1.42 m to the table) | **Exteroceptive sensor** | Measures distances to obstacles and surfaces. | **Sense → Perceive:** used to locate obstacles and the table for safe navigation and for the hand-over position. |
| 7 | **Base** and **torso** | **Base, torso** | The base carries everything. The torso links base, arms and head, and can adjust the working height. | **Act / Propose:** the torso height is a degree of freedom for the grasp pose and the hand-over height. |
| 8 | **Head** | **Head** | Carries the cameras and points them at the task. | **Perceive:** controls where the robot looks. |

**Sensor types and where they sit in our code:**

- **Exteroception** (sensing the outside world): the camera and the laser scanner. In the code this is `SensingSystem.sense`, which measures temperature, friction and Eva's position.
- **Proprioception** (sensing the robot's own body): joint encoders and force/torque sensors in the motors, arms and gripper. They are not drawn in the figure, but they let the robot check that its movement follows the plan.
- **Interoception** (sensing the robot's own state): the battery level and motor temperature. We use this in Task 4, where the `energetically_viable` check turns the battery level into a decision.

**Summary:** the sensors (red) feed the `Observation`, the actuators and the gripper (blue) carry out the chosen `ParameterPacket`, and the links and joints decide which motions are possible. The Warrant Gate decides whether the robot may start the motion at all.
